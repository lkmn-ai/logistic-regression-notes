## Evaluation Metrics for Classification

```python
Confusion Matrix :
    - It is a N * N matrix used to evaluate performance of Classification Model
    - It compares actual values and predicted values

                           Actual Values
                        Positive      Negative
Predicted  Positive        TP           FP
Values     Negative        FN           TN

    ex. Positive, Negative
    - True Positive (TP):
        - Actual value is Positive and Predicted as Positive
        - the pred(Positive) matches with actual (Positive) value.

    - True Negative (TN):
        - Actual value is Negative and Predicted as Negative
        - the pred(Negative) matches with actual (Negative) value.

    - False Positive (FP): Type-1 Error
        - Actual value is Negative and Predicted as Positive
        - the pred(Positive) not matches with actual (Negative) value.
        - the pred value is falsely predicted

    - False Negative (FN): Type-2 Error
        - Actual value is Positive and Predicted as Negative
        - the pred(Negative) not matches with actual (Positive) value.
        - the pred value is falsely predicted
```

```python
Accuracy:
    - It is the ratio of total number of correctly classified values to the total number of observations

    - Accuracy >> TP + TN / (TP+TN+FN+FP)
    - Accuracy ranges from 0 to 1
    - Accuracy only counts the correct predictions, which is why the numerator is TP + TN.


>> When data is balanced, this is the most appropriate metric to use
>> When data is imbalanced, accuracy may not perform better, mostly in case of multi-class classifier
```

```python
Precision:

    Precision = TP/TP+FP

    We will focus on FP
    It ranges from 0 to 1

    Out of total predicted positive results(TP+FP) how many are actually positive(TP)

    When FP is important then Precision is the appropriate metric
    When money is on stake, Stock-Market, E-commerce, Finance,Banking
```

```python
Recall:

    Recall = TP/TP+FN

    We will focus on FN
    It ranges from 0 to 1

    Out of total actually positive results(TP+FN) how many are predicted positive(TP)

    When FN is important then Recall is the appropriate metric
    When lives are on stake, Medical, Pharma
```

```python
F1-Score:
    f1-score = 2PR/P+R

    It is harmonic mean of Precision and Recall
    When Precision is low and Recall is high or vice-versa, its difficult to compare
    In such a cases f1-score is a good metric to use
    It strikes the balance between Precision and Recall


Accuracy → "How often is my model correct overall?"
Precision → "When my model predicts Positive, can I trust it?"
Recall → "Did my model find most of the actual Positive cases?"
F1-Score → "Is my model good at both Precision and Recall, or is it only strong in one?"
```

## AUC ROC Curve

```python
ROC : Reciever Operating Characteristics
AUC : Area Under Curve
```

```python
True Positive Rate/Recall/Sensitivity:

    TPR = TP/(TP+FN)

    What proportion of positive class got correctly classified by the classifier

False Negative Rate:

    FNR = FN/(TP+FN)

    What proportion of positive class got incorrectly classified by the classifier

True Negative Rate/Specificity:

    TNR = TN/(TN+FP)

    What proportion of negative class got correctly classified by the classifier

False Positive Rate/(1- Specificity):

    FPR = FP/(TN+FP)

    What proportion of negative class got incorrectly classified by the classifier


Better TPR and TNR, Better the model

Higher FNR and FPR, Worst the model

The main aim is to maximixe the area under the curve
```

```python
TNR/Specificity = 0.81

FPR(1 - Specificity) = 0.19 / (1-0.81)
```
